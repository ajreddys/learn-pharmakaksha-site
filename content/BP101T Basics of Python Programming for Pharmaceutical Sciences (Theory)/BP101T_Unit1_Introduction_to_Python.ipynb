## 1.3 Variables and data types
A **variable** is a name for a value, like a labelled bottle on a pharmacy shelf. Anything after `#` is a **comment**. Python ignores it.

```python
drug_name = "Paracetamol"      # str
tablets_per_strip = 10         # int
dose_mg = 500.0                # float
is_otc = True                  # bool

print(type(drug_name), type(tablets_per_strip), type(dose_mg), type(is_otc))
```

**Naming rules.** A name cannot start with a digit. Remove the `#` from the next line and run it to see the error, then put the `#` back.

```python
# 1dose = 5        # SyntaxError: a name cannot start with a digit
dose1 = 5          # this is fine
Dose = 10          # names are case-sensitive: Dose and dose1 are different variables
print(dose1, Dose)
```

**The four basic data types**

```python
tablets = 30              # int: whole numbers
cmax = 18.4               # float: decimal numbers
drug = "Metformin"        # str: text in quotes
is_schedule_h = True      # bool: True or False

print(tablets, cmax, drug, is_schedule_h)
print(type(tablets), type(cmax), type(drug), type(is_schedule_h))
```

**Multiple assignment** puts values into several variables in one line:

```python
weight, height = 62.5, 170
print(weight, height)
```

**Common mistake:** a number inside quotes is text. Multiplying text repeats it.

```python
dose = "500"
print(dose * 2)      # 500500, not 1000!
print(type(dose))    # <class 'str'>
```

## 1.4 Type casting and operators
**Type casting** converts a value from one type to another.

```python
print(int("250"))      # 250    text to integer
print(float("12.5"))   # 12.5   text to decimal
print(str(500))        # 500    number to text (now a str)
print(int(7.9))        # 7      truncates, does NOT round
print(bool(0))         # False  0 and empty values are False
print(bool("no"))      # True   any non-empty text is True
```

`int("12.5")` raises a **ValueError**. Remove the `#` to see it, then put it back. The fix is to convert with `float()` first.

```python
# int("12.5")                 # ValueError
print(int(float("12.5")))     # 12
```

**Arithmetic operators**

```python
print(250 + 250)     # addition        500
print(100 - 18.4)    # subtraction     81.6
print(62.5 * 15)     # multiplication  937.5  (62.5 kg x 15 mg/kg)
print(24 / 6)        # division        4.0    (always a float)
print(24 // 6)       # floor division  4
print(25 % 6)        # modulus         1      (remainder)
print(2 ** 3)        # power           8
```

A drug given every 6 hours is taken `24 / 6` = 4.0 times a day. Use `//` when you need a whole number of doses.

**Comparison operators** give `True` or `False`:

```python
cmax = 18.4
mtc = 20              # minimum toxic concentration
print(cmax < mtc)     # True, below the toxic level
print(cmax == mtc)    # False
```

**Logical operators** combine conditions. The patient is 65 years old with a creatinine clearance (CrCl) of 45 mL/min:

```python
age, crcl = 65, 45
is_otc = True
print(age >= 60 and crcl < 50)   # True: both hold
print(age < 18 or crcl < 30)     # False: neither holds
print(not is_otc)                # False
```

## 1.5 Input and output
`print()` shows values. Items separated by commas print with a space between them.

```python
weight = 62.5
dose = weight * 15
print("Dose for", weight, "kg =", dose, "mg")
print(f"Dose: {dose:.1f} mg")          # f-string, one decimal place
```

`sep` changes the separator, `end` replaces the new line, `\n` starts a new line and `\t` inserts a tab:

## 1.6 String operations
Each character has an **index** starting at 0. Negative indices count from the end.

```python
batch = "BATCH-2026-0457"
print(batch[0])        # B      first character
print(batch[-1])       # 7      last character
print(batch[0:5])      # BATCH  index 0 up to (not including) 5
print(batch[-4:])      # 0457   last four characters
print(len(batch))      # 15
```

```python
print("Para" + "cetamol")      # join
print("-" * 20)                # repeat
print("2026" in batch)         # membership: True
```

**String methods** return a new string; the original never changes.

```python
print("amoxicillin".upper())
print("amoxicillin trihydrate".title())
print("  aspirin  ".strip())
print(batch.replace("-", "/"))
print(batch.split("-"))
print(batch.startswith("BATCH"))
print(batch.find("2026"))
print("B.Pharm".count("a"))
```

**Why it matters:** the same drug typed three ways becomes one clean name.

```python
print(" PARACETAMOL".strip().title())
print("paracetamol ".strip().title())
print("Paracetamol".strip().title())
```

## 1.7 Libraries
**Standard libraries** such as `math` come with Python.

```python
import math

k = 0.173                                  # elimination rate constant (per hr)
t_half = math.log(2) / k                   # t½ = 0.693 / k
print(f"Half-life = {t_half:.2f} h")

bsa = math.sqrt(170 * 65 / 3600)           # Mosteller BSA for 170 cm, 65 kg
print(f"BSA = {bsa:.2f} m²")
```

## Worked example: paediatric syrup dose calculator
This program uses the whole unit: input, type casting, string methods, operators, f-strings and `min()`.

⌨️ Run it and type: `paracetamol`, `18`, `15`, `1000`, `250`. You should get 270.0 mg and 5.4 mL.

```python
print("=" * 40)
print("PHARMAKAKSHA PAEDIATRIC DOSE CALCULATOR")
print("=" * 40)

drug = input("Drug name: ").strip().title()            # string methods
weight = float(input("Body weight (kg): "))            # type casting
dose_per_kg = float(input("Dose (mg/kg per dose): "))
max_dose = float(input("Maximum single dose (mg): "))
strength = float(input("Syrup strength (mg per 5 mL): "))

calculated = weight * dose_per_kg                      # arithmetic
dose = min(calculated, max_dose)                       # never exceed the maximum
capped = calculated > max_dose                         # comparison gives a bool
volume_ml = dose / strength * 5

print("-" * 40)
print(f"Drug          : {drug}")
print(f"Single dose   : {dose:.1f} mg")
print(f"Volume        : {volume_ml:.1f} mL")
print(f"Capped at max : {capped}")
```

The same calculation with the sample values typed in, so you can check your answer:

```python
drug, weight, dose_per_kg, max_dose, strength = "Paracetamol", 18, 15, 1000, 250

calculated = weight * dose_per_kg
dose = min(calculated, max_dose)
capped = calculated > max_dose
volume_ml = dose / strength * 5

print(f"Drug          : {drug}")
print(f"Single dose   : {dose:.1f} mg")
print(f"Volume        : {volume_ml:.1f} mL")
print(f"Capped at max : {capped}")
```
