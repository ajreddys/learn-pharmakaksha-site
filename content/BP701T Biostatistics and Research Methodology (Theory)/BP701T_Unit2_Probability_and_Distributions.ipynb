## 2.1–2.3 Probability rules on a two-way table
200 patients: Drug A (n = 100) had 18 ADRs; Drug B (n = 100) had 8.

```python
t = pd.DataFrame({"ADR": [18, 8], "No ADR": [82, 92]}, index=["Drug A", "Drug B"])
t.loc["Total"] = t.sum(); t["Total"] = t.sum(axis=1)
print(t, "\n")
N = 200
P_ADR = 26 / N; P_A = 100 / N; P_A_and_ADR = 18 / N
print("Marginal  P(ADR)        =", P_ADR)
print("Joint     P(A and ADR)  =", P_A_and_ADR)
print("Addition  P(A or ADR)   =", P_A + P_ADR - P_A_and_ADR)
print("Conditional P(ADR | A)  =", P_A_and_ADR / P_A, "| P(ADR | B) =", 8 / 100)
print("Independent?", np.isclose(P_A_and_ADR, P_A * P_ADR))
print("Two defectives from 20 tablets with 3 defective (no replacement):", round(3/20 * 2/19, 4))
```

### Bayes' theorem: a screening test
Sensitivity 95%, specificity 90%, prevalence 2%. What is P(disease | positive)?

```python
se, sp, prev = 0.95, 0.90, 0.02
tp = se * prev; fp = (1 - sp) * (1 - prev)
print("P(D | +) =", round(tp / (tp + fp), 3))
```

## 2.5 Binomial distribution
Cure rate p = 0.8; 6 patients treated. Hand answers: P(X = 6) = 0.262, P(X = 5) = 0.393, P(X ≥ 5) = 0.655, mean 4.8, variance 0.96.

```python
from math import comb
n, p = 6, 0.8
for k in range(n + 1):
    print(k, f"{comb(n, k) * p**k * (1-p)**(n-k):.4f}", f"{stats.binom.pmf(k, n, p):.4f}")
print("P(X >= 5) =", round(stats.binom.sf(4, n, p), 4), "| mean", n*p, "| variance", round(n*p*(1-p), 2))
plt.bar(range(7), stats.binom.pmf(range(7), n, p), color="#1F8A70"); plt.title("Binomial PMF, n = 6, p = 0.8"); plt.xlabel("patients cured"); plt.show()
```

## 2.6 Poisson distribution
A hospital receives on average λ = 2 reports of a rare ADR per month. Hand answers: P(0) = 0.135, P(≤ 2) = 0.677, P(≥ 3) = 0.323.

```python
from math import exp, factorial
lam = 2
for k in range(5):
    print(k, f"{exp(-lam) * lam**k / factorial(k):.4f}", f"{stats.poisson.pmf(k, lam):.4f}")
print("P(X <= 2) =", round(stats.poisson.cdf(2, lam), 4), "| P(X >= 3) =", round(stats.poisson.sf(2, lam), 4))
print("Poisson approximation to binomial (n = 2000, p = 0.001): exact P(0) =", round(stats.binom.pmf(0, 2000, 0.001), 4))
```

## 2.7 Normal distribution and Z
Tablet weight ~ N(μ = 250, σ = 5). Hand answers: P(X < 240) = 0.0228; P(245 < X < 255) = 0.6827; P(237.5 < X < 262.5) = 0.9876.

```python
mu, sigma = 250, 5
z = (240 - mu) / sigma
print("Z =", z, "| P(X < 240) =", round(stats.norm.cdf(z), 4))
print("P(245 < X < 255) =", round(stats.norm.cdf(255, mu, sigma) - stats.norm.cdf(245, mu, sigma), 4))
print("Within ±5% of label =", round(stats.norm.cdf(262.5, mu, sigma) - stats.norm.cdf(237.5, mu, sigma), 4))
print("Z-table area 0 to 1.5 =", round(stats.norm.cdf(1.5) - 0.5, 4))
xs = np.linspace(230, 270, 300)
plt.plot(xs, stats.norm.pdf(xs, mu, sigma), color="#12303A")
plt.fill_between(xs, stats.norm.pdf(xs, mu, sigma), where=(xs >= 245) & (xs <= 255), color="#1F8A70", alpha=.5, label="μ ± 1σ = 68.3%")
plt.legend(); plt.title("Tablet weight ~ N(250, 5²)"); plt.show()
```

## 2.8 Reading t, χ² and F tables
The table value is the **critical value** that leaves α in the tail. Compare with your printed table.

```python
print("t (two-tailed 0.05), df 9  :", round(stats.t.ppf(0.975, 9), 3))
print("t (two-tailed 0.05), df 14 :", round(stats.t.ppf(0.975, 14), 3))
print("Z (two-tailed 0.05)        :", round(stats.norm.ppf(0.975), 3))
print("chi-square 0.05, df 1      :", round(stats.chi2.ppf(0.95, 1), 3))
print("chi-square 0.05, df 2      :", round(stats.chi2.ppf(0.95, 2), 3))
print("F 0.05 (2, 12)             :", round(stats.f.ppf(0.95, 2, 12), 3))
xs = np.linspace(-4, 4, 300)
plt.plot(xs, stats.norm.pdf(xs), label="Z (normal)")
for df in (2, 9): plt.plot(xs, stats.t.pdf(xs, df), "--", label=f"t, df = {df}")
plt.legend(); plt.title("t has heavier tails than Z; it approaches Z as df grows"); plt.show()
```

## Worked example · Will this batch pass the weight test?
Notes, worked pharma example. Hand answers: P(outside ±5%) = 0.0124; P(batch fails) ≈ 0.0021 (Poisson) / 0.0019 (binomial); with σ = 8 mg, 0.43.

```python
p = 2 * stats.norm.sf(2.5)
print("P(one tablet outside +/-5%) =", round(p, 4))
lam = 20 * p
for k in range(3):
    print(k, "Poisson", round(stats.poisson.pmf(k, lam), 4), "| binomial", round(stats.binom.pmf(k, 20, p), 4))
print("P(fail) Poisson", round(stats.poisson.sf(2, lam), 4), "| binomial", round(stats.binom.sf(2, 20, p), 4))
p8 = 2 * stats.norm.sf(12.5 / 8)
print("sigma = 8: P(outside) =", round(p8, 4), "| P(batch fails) =", round(stats.binom.sf(2, 20, p8), 2))
```
